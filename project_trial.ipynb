```python
import imdlib as imd

# Hum current horizon (jaise 2025) ka data download kar rahe hain
start_yr = 2025
end_yr = 2025
variable = 'rain' 

# Data download process
data = imd.get_data(variable, start_yr, end_yr, fn_format='nc')

# Isko local directory mein save kar lo
data.to_netcdf('india_rain_2025.nc')
print("IMD Data download hokar save ho gaya hai!")
```
